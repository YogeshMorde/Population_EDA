# 🌍 Task 13: Find Top 5 Fastest Growing Countries

```python
df
```

```python
df.dtypes
```

```python
import pandas as pd

#df = pd.read_csv("population.csv")

# Clean up: remove % and normalize minus signs
df['Decadal Growth (%)'] = (
    df['Decadal Growth (%)']
    .str.replace('%','')
    .str.replace('−','-')   # replace special minus with normal hyphen
    .astype(float)
)
```

```python
df.dtypes
```

```python
df.plot(x='State', y='Decadal Growth (%)', kind='bar', figsize=(12,6))
```

# 📈 Task 14: Create Population Growth Trend Line Chart

```python
plt.figure(figsize=(12,6))
plt.plot(df['State'], df['Decadal Growth (%)'], marker='o')
plt.title("Population Growth Trend by State")
plt.xlabel("State")
plt.ylabel("Decadal Growth (%)")
plt.xticks(rotation=90)
plt.show()
```

# 🗺️ Task 15: Build Choropleth Map

```python
import plotly.express as px

fig = px.choropleth(
    df,
    locations='State',  # Fix: Changed from 'Country' to 'State'
    locationmode='country names', 
    color='Population'
)

# Note: Use fig.show() for Plotly charts. plt.show() belongs to Matplotlib.
fig.show()
```
